```python
%%writefile /content/work/make_paired_swaps.py
"""Build a PAIRED real/fake dataset (real image + same image with another face swapped in)."""
import argparse
import csv
import random
import time
import zlib
from pathlib import Path

import cv2
import numpy as np

EXTENSIONS = {".jpg", ".jpeg", ".png", ".webp", ".bmp"}


def read_image(path):
    data = np.fromfile(str(path), dtype=np.uint8)
    return cv2.imdecode(data, cv2.IMREAD_COLOR)


def write_png(path, image):
    path.parent.mkdir(parents=True, exist_ok=True)
    ok, encoded = cv2.imencode(".png", image)
    if not ok:
        raise RuntimeError(f"Could not encode {path}")
    encoded.tofile(str(path))


def upscale(image):
    h, w = image.shape[:2]
    pad = int(0.5 * max(h, w))
    padded = cv2.copyMakeBorder(image, pad, pad, pad, pad, cv2.BORDER_CONSTANT, value=(127, 127, 127))
    ph, pw = padded.shape[:2]
    scale = max(1.0, 640.0 / min(ph, pw))
    if scale > 1.0:
        padded = cv2.resize(padded, (int(round(pw * scale)), int(round(ph * scale))), interpolation=cv2.INTER_CUBIC)
    return padded, pad, (h, w, ph, pw)


def restore(big, pad, dims):
    h, w, ph, pw = dims
    if big.shape[:2] != (ph, pw):
        big = cv2.resize(big, (pw, ph), interpolation=cv2.INTER_AREA)
    return big[pad:pad + h, pad:pad + w]


def assign_split(relative_name, test_share, val_share):
    bucket = zlib.crc32(relative_name.encode("utf-8")) % 1000
    if bucket < test_share * 1000:
        return "test"
    if bucket < (test_share + val_share) * 1000:
        return "val"
    return "train"


def largest_face(app, image):
    faces = app.get(image)
    if not faces:
        return None
    return max(faces, key=lambda f: (f.bbox[2] - f.bbox[0]) * (f.bbox[3] - f.bbox[1]))


def main():
    parser = argparse.ArgumentParser()
    parser.add_argument("real_folder", type=Path)
    parser.add_argument("output_folder", type=Path)
    parser.add_argument("--count", type=int, default=1500)
    parser.add_argument("--model", default="inswapper_128.onnx")
    parser.add_argument("--val-share", type=float, default=0.15)
    parser.add_argument("--test-share", type=float, default=0.15)
    parser.add_argument("--seed", type=int, default=42)
    parser.add_argument("--gpu", action="store_true")
    args = parser.parse_args()

    import insightface
    from insightface.app import FaceAnalysis

    paths = sorted(p for p in args.real_folder.rglob("*") if p.is_file() and p.suffix.lower() in EXTENSIONS)
    if not paths:
        raise SystemExit(f"No images found in {args.real_folder}")

    rng = random.Random(args.seed)
    rng.shuffle(paths)
    paths = paths[: args.count]

    splits = {"test": [], "val": [], "train": []}
    for path in paths:
        name = str(path.relative_to(args.real_folder))
        splits[assign_split(name, args.test_share, args.val_share)].append(path)

    providers = ["CUDAExecutionProvider", "CPUExecutionProvider"] if args.gpu else ["CPUExecutionProvider"]
    app = FaceAnalysis(name="buffalo_l", providers=providers)
    app.prepare(ctx_id=0 if args.gpu else -1, det_size=(640, 640))
    try:
        swapper = insightface.model_zoo.get_model(args.model, providers=providers)
    except TypeError:
        swapper = insightface.model_zoo.get_model(args.model)

    manifest_path = args.output_folder / "manifest.csv"
    args.output_folder.mkdir(parents=True, exist_ok=True)
    manifest = open(manifest_path, "a", newline="", encoding="utf-8")
    writer = csv.writer(manifest)
    if manifest_path.stat().st_size == 0:
        writer.writerow(["split", "real_file", "fake_file", "target", "donor"])

    started = time.time()
    total_made = 0
    total_skipped = 0

    for split, split_paths in splits.items():
        print(f"\n[{split}] finding faces in {len(split_paths)} images...")
        faces = {}
        for path in split_paths:
            image = read_image(path)
            if image is None:
                continue
            big, _, _ = upscale(image)
            face = largest_face(app, big)
            if face is not None:
                faces[path] = face
        valid = sorted(faces.keys())
        total_skipped += len(split_paths) - len(valid)
        print(f"[{split}] {len(valid)} usable, {len(split_paths) - len(valid)} skipped (no face)")

        if len(valid) < 2:
            print(f"[{split}] not enough images, skipping this split")
            continue

        for target_path in valid:
            stem = f"{split}_{target_path.stem}"
            real_out = args.output_folder / split / "real" / f"{stem}.png"
            fake_out = args.output_folder / split / "fake" / f"{stem}_swap.png"
            if real_out.exists() and fake_out.exists():
                continue

            donor_path = rng.choice([p for p in valid if p != target_path])
            image = read_image(target_path)
            big, pad, dims = upscale(image)
            swapped = swapper.get(big, faces[target_path], faces[donor_path], paste_back=True)

            write_png(real_out, restore(big, pad, dims))
            write_png(fake_out, restore(swapped, pad, dims))
            writer.writerow([split, real_out.name, fake_out.name, target_path.name, donor_path.name])
            manifest.flush()
            total_made += 1

            if total_made % 25 == 0:
                print(f"  made {total_made} pairs ({(time.time() - started) / 60:.1f} min)")

    manifest.close()
    print(f"\nFinished: {total_made} new pairs, {total_skipped} skipped. Output: {args.output_folder}")


if __name__ == "__main__":
    main()
```

```python
%%writefile /content/work/prepare_data.py
"""Turn raw images into model-ready 224x224 face crops using the SAME steps as backend.py:
YuNet face -> crop with 35% margin -> pad to square -> Lanczos resize to 224."""
import csv
from pathlib import Path

import cv2
import numpy as np
from PIL import Image, ImageOps

WORK = Path("/content/work")
OUT = WORK / "data_ready"
YUNET_MODEL_PATH = str(WORK / "face_detection_yunet_2023mar.onnx")
YUNET_SCORE_THRESHOLD = 0.80
YUNET_NMS_THRESHOLD = 0.3
YUNET_TOP_K = 5000
YUNET_MAX_DETECTION_SIZE = 640
MODEL_INPUT_SIZE = (224, 224)
FACE_CROP_MARGIN = 0.35
EXTENSIONS = {".jpg", ".jpeg", ".png", ".webp", ".bmp"}

# Images that belong to the team's shared test set. Anything made from them
# (as the picture OR as the donor face) is kept out of training.
SHARED_TEST_STEMS = {
    "real_9940", "real_9941", "real_9942", "real_9943", "real_9944", "real_9945", "real_9946",
    "real_9947", "real_9948", "real_9949", "real_995", "real_9950", "real_9951", "real_9952",
    "real_9953", "real_9954", "real_9955", "real_9956", "real_9957", "real_9958", "real_9959",
    "real_996", "real_9960", "real_9961", "real_9962", "real_9963", "real_9964", "real_9965",
    "real_9966", "real_9967", "real_9968", "real_9969", "real_997", "real_9970", "real_9971",
    "real_9972", "real_9973", "real_9974", "real_9975", "real_9976", "real_9977", "real_9978",
    "real_9979", "real_998", "real_9980", "real_9981", "real_9982", "real_9983", "real_9984",
    "real_999",
}

_detector = None


def get_detector():
    global _detector
    if _detector is None:
        _detector = cv2.FaceDetectorYN.create(
            YUNET_MODEL_PATH, "", (320, 320),
            YUNET_SCORE_THRESHOLD, YUNET_NMS_THRESHOLD, YUNET_TOP_K,
        )
    return _detector


def detect_face(image):
    """Same logic as backend.py detect_human_face. Returns a face box dict or None."""
    detector = get_detector()
    bgr = cv2.cvtColor(np.asarray(image), cv2.COLOR_RGB2BGR)
    original_height, original_width = bgr.shape[:2]
    scale = min(1.0, YUNET_MAX_DETECTION_SIZE / float(max(original_width, original_height)))
    if scale < 1.0:
        bgr = cv2.resize(
            bgr,
            (max(1, int(original_width * scale)), max(1, int(original_height * scale))),
            interpolation=cv2.INTER_AREA,
        )
    height, width = bgr.shape[:2]
    detector.setInputSize((width, height))
    _, faces = detector.detect(bgr)
    if faces is None or len(faces) == 0:
        return None
    face = max(faces, key=lambda f: float(f[2] * f[3] * f[14]))
    inverse = 1.0 / scale
    x = max(0, min(int(round(float(face[0]) * inverse)), original_width - 1))
    y = max(0, min(int(round(float(face[1]) * inverse)), original_height - 1))
    w = max(1, min(int(round(float(face[2]) * inverse)), original_width - x))
    h = max(1, min(int(round(float(face[3]) * inverse)), original_height - y))
    return {"x": x, "y": y, "width": w, "height": h}


def crop_face(image, box, margin=FACE_CROP_MARGIN):
    width, height = image.size
    mx, my = int(box["width"] * margin), int(box["height"] * margin)
    return image.crop((
        max(0, box["x"] - mx), max(0, box["y"] - my),
        min(width, box["x"] + box["width"] + mx), min(height, box["y"] + box["height"] + my),
    ))


def to_model_image(image):
    width, height = image.size
    size = max(width, height)
    square = Image.new("RGB", (size, size), (0, 0, 0))
    square.paste(image, ((size - width) // 2, (size - height) // 2))
    return square.resize(MODEL_INPUT_SIZE, Image.Resampling.LANCZOS)


def clean_stem(file_name):
    """train_real_9940_swap.png -> real_9940"""
    stem = Path(file_name).stem
    for prefix in ("train_", "val_", "test_"):
        if stem.startswith(prefix):
            stem = stem[len(prefix):]
    if stem.endswith("_swap"):
        stem = stem[:-5]
    return stem


def is_shared_test(name):
    return Path(name).stem.lower() in SHARED_TEST_STEMS


def convert(files, out_dir, prefix):
    made, skipped = 0, 0
    out_dir.mkdir(parents=True, exist_ok=True)
    for path in files:
        try:
            image = ImageOps.exif_transpose(Image.open(path)).convert("RGB")
        except Exception:
            skipped += 1
            continue
        box = detect_face(image)
        if box is None:
            skipped += 1
            continue
        to_model_image(crop_face(image, box)).save(out_dir / f"{prefix}{path.stem}.png")
        made += 1
    return made, skipped


def list_images(folder):
    if not folder.exists():
        return []
    return sorted(p for p in folder.rglob("*") if p.is_file() and p.suffix.lower() in EXTENSIONS)


def paired_sources(split):
    """(real_files, fake_files) for one split, dropping pairs that touch the shared test set."""
    base = WORK / "paired_data"
    manifest = {}
    manifest_path = base / "manifest.csv"
    if manifest_path.exists():
        with open(manifest_path, newline="", encoding="utf-8") as file:
            for row in csv.DictReader(file):
                if row.get("split") == split:
                    manifest[row["fake_file"]] = row
    real_files, fake_files, dropped = [], [], 0
    for fake in list_images(base / split / "fake"):
        row = manifest.get(fake.name)
        target = Path(row["target"]).stem.lower() if row else clean_stem(fake.name).lower()
        donor = Path(row["donor"]).stem.lower() if row else ""
        if target in SHARED_TEST_STEMS or donor in SHARED_TEST_STEMS:
            dropped += 1
            continue
        real = base / split / "real" / (fake.name[: -len("_swap.png")] + ".png")
        if real.exists():
            real_files.append(real)
            fake_files.append(fake)
    return real_files, fake_files, dropped


def main():
    if not Path(YUNET_MODEL_PATH).exists():
        raise SystemExit(f"YuNet model not found: {YUNET_MODEL_PATH}")

    jobs = []  # (split, class, files, prefix)
    for split in ("train", "val", "test"):
        real_files, fake_files, dropped = paired_sources(split)
        print(f"paired_data/{split}: {len(real_files)} pairs kept, {dropped} dropped (touch shared test set)")
        jobs.append((split, "real", real_files, "pd_"))
        jobs.append((split, "fake", fake_files, "pd_"))

    kaggle = WORK / "kaggle_subset"
    for split in ("train", "val"):
        for source_class, label in (("original", "real"), ("altered", "fake")):
            files = list_images(kaggle / split / source_class)
            print(f"kaggle_subset/{split}/{source_class}: found {len(files)} files")
            jobs.append((split, label, files, "kg_"))

    if not any(files for _, _, files, _ in jobs):
        raise SystemExit("No images found anywhere. Check the copy/unzip step.")

    total_skipped = 0
    for split, label, files, prefix in jobs:
        made, skipped = convert(files, OUT / split / label, prefix)
        total_skipped += skipped
        print(f"  -> {split}/{label} from {prefix[:-1]}: {made} saved, {skipped} skipped")

    print("\nFINAL COUNTS in data_ready:")
    for split in ("train", "val", "test"):
        for label in ("real", "fake"):
            print(f"  {split}/{label}: {len(list_images(OUT / split / label))}")
    print(f"skipped in total: {total_skipped}")


if __name__ == "__main__":
    main()
```

```python
%%writefile /content/work/train.py
"""Fine-tune Detect Now on Kaggle + inswapper data (run in Colab).

Reads data_ready/{train,val,test}/{real,fake} made by prepare_data.py.
File names starting kg_ come from Kaggle, pd_ from the inswapper pairs.
The threshold is chosen on VAL only. TEST (inswapper pairs the model never saw)
is used once, at the end, for the honest number.
"""
import json
import os
import shutil
from pathlib import Path

os.environ["TF_CPP_MIN_LOG_LEVEL"] = "2"

import numpy as np

WORK = Path("/content/work")
DATA = WORK / "data_ready"
BASE_MODEL = WORK / "detect_now_efficientnetb0_faceswap_crops.keras"
STEM = "detect_now_efficientnetb0_faceswap_crops_v2"
OUT_MODEL = WORK / f"{STEM}.keras"
OUT_THRESHOLD = WORK / f"decision_threshold_{STEM}.json"
DRIVE_OUT = Path("/content/drive/MyDrive/detect_now_v2")

SEED = 42
BATCH = 32
EPOCHS = 12
LEARNING_RATE = 2e-5
UNFREEZE_LAST = 100          # backbone layers that may change (batch-norm stays frozen)
MAX_FALSE_POSITIVE_RATE = 0.05
CLASS_NAMES = ["real", "fake"]   # sigmoid output: 0 = real, 1 = fake


def auc_score(real_scores, fake_scores):
    fake, real = fake_scores[:, None], real_scores[None, :]
    return float((fake > real).mean() + 0.5 * (fake == real).mean())


def metrics_at(labels, probs, threshold):
    pred = probs >= threshold
    tp = int(np.sum(pred & (labels == 1)))
    fp = int(np.sum(pred & (labels == 0)))
    tn = int(np.sum(~pred & (labels == 0)))
    fn = int(np.sum(~pred & (labels == 1)))
    precision = tp / max(tp + fp, 1)
    recall = tp / max(tp + fn, 1)
    return {
        "threshold": round(float(threshold), 2),
        "real_correct": tn / max(tn + fp, 1),
        "fakes_caught": recall,
        "accuracy": (tp + tn) / max(len(labels), 1),
        "false_positive_rate": fp / max(fp + tn, 1),
        "f1": 2 * precision * recall / max(precision + recall, 1e-8),
    }


def choose_threshold(labels, probs):
    """Best F1 while real images wrongly flagged stay under the cap."""
    best_any, best_capped = None, None
    for t in np.arange(0.05, 0.96, 0.01):
        m = metrics_at(labels, probs, t)
        if best_any is None or m["f1"] > best_any["f1"]:
            best_any = m
        if m["false_positive_rate"] <= MAX_FALSE_POSITIVE_RATE:
            if best_capped is None or m["f1"] > best_capped["f1"]:
                best_capped = m
    return best_capped or best_any


def subset_auc(labels, probs, mask):
    labels, probs = labels[mask], probs[mask]
    if (labels == 0).sum() == 0 or (labels == 1).sum() == 0:
        return float("nan")
    return auc_score(probs[labels == 0], probs[labels == 1])


def find_backbone(tf, model):
    try:
        return model.get_layer("efficientnetb0")
    except ValueError:
        pass
    for layer in model.layers:
        if isinstance(layer, tf.keras.Model):
            return layer
    raise ValueError("Could not find the EfficientNet backbone: " + ", ".join(l.name for l in model.layers))


def load_split(tf, split, **kwargs):
    return tf.keras.utils.image_dataset_from_directory(
        str(DATA / split), labels="inferred", label_mode="binary", class_names=CLASS_NAMES,
        image_size=(224, 224), seed=SEED, **kwargs,
    )


def score_split(tf, model, split):
    ds = load_split(tf, split, batch_size=64, shuffle=False)
    paths = list(ds.file_paths)
    labels = np.concatenate([y.numpy().reshape(-1) for _, y in ds]).astype(int)
    probs = np.asarray(model.predict(ds, verbose=0)).reshape(-1)
    return paths, labels, probs


def augment_fn(tf):
    def augment(image, label):
        image = tf.image.random_flip_left_right(image)

        def jpeg():
            small = tf.cast(tf.clip_by_value(image, 0.0, 255.0), tf.uint8)
            small = tf.image.random_jpeg_quality(small, 45, 95)
            return tf.cast(small, tf.float32)

        image = tf.cond(tf.random.uniform([]) < 0.5, jpeg, lambda: image)
        image = tf.image.random_brightness(image, 20.0)
        image = tf.image.random_contrast(image, 0.85, 1.15)
        image = tf.clip_by_value(image, 0.0, 255.0)
        image.set_shape([224, 224, 3])
        return image, label
    return augment


def report(title, paths, labels, probs, threshold=None):
    is_kaggle = np.array([Path(p).name.startswith("kg_") for p in paths])
    print(f"\n{title}")
    for name, mask in (("Kaggle ", is_kaggle), ("inswapper", ~is_kaggle)):
        if mask.sum() == 0:
            continue
        line = f"  {name:9s} n={int(mask.sum()):4d}  AUC={subset_auc(labels, probs, mask):.3f}"
        if threshold is not None:
            m = metrics_at(labels[mask], probs[mask], threshold)
            line += (f"  @ {threshold:.2f}: real correct {m['real_correct']:.1%}, "
                     f"fakes caught {m['fakes_caught']:.1%}, accuracy {m['accuracy']:.1%}")
        print(line)


def main():
    import tensorflow as tf

    tf.keras.utils.set_random_seed(SEED)
    print("TensorFlow", tf.__version__, "| Keras", tf.keras.__version__)
    print("GPUs:", tf.config.list_physical_devices("GPU"))

    # ---- data ----
    train = (load_split(tf, "train", batch_size=None, shuffle=False)
             .cache()
             .shuffle(4096, seed=SEED, reshuffle_each_iteration=True)
             .map(augment_fn(tf), num_parallel_calls=tf.data.AUTOTUNE)
             .batch(BATCH).prefetch(tf.data.AUTOTUNE))
    val = load_split(tf, "val", batch_size=BATCH, shuffle=False).cache().prefetch(tf.data.AUTOTUNE)

    # ---- "before" numbers for the model as it is today ----
    old_model = tf.keras.models.load_model(str(BASE_MODEL), compile=False)
    v_paths, v_labels, old_val = score_split(tf, old_model, "val")
    t_paths, t_labels, old_test = score_split(tf, old_model, "test")
    report("BEFORE (current model) - validation", v_paths, v_labels, old_val, 0.45)
    report("BEFORE (current model) - test", t_paths, t_labels, old_test, 0.45)
    del old_model

    # ---- fine-tune ----
    model = tf.keras.models.load_model(str(BASE_MODEL), compile=False)
    model.trainable = True
    backbone = find_backbone(tf, model)
    backbone.trainable = True
    for layer in backbone.layers[:-UNFREEZE_LAST]:
        layer.trainable = False
    for layer in backbone.layers[-UNFREEZE_LAST:]:
        layer.trainable = not isinstance(layer, tf.keras.layers.BatchNormalization)
    print("trainable weights:", sum(int(np.prod(w.shape)) for w in model.trainable_weights))

    model.compile(
        optimizer=tf.keras.optimizers.Adam(LEARNING_RATE),
        loss=tf.keras.losses.BinaryCrossentropy(),
        metrics=[tf.keras.metrics.BinaryAccuracy(name="accuracy"), tf.keras.metrics.AUC(name="auc")],
    )
    model.fit(
        train, validation_data=val, epochs=EPOCHS, verbose=2,
        callbacks=[
            tf.keras.callbacks.ModelCheckpoint(str(OUT_MODEL), monitor="val_auc", mode="max", save_best_only=True),
            tf.keras.callbacks.EarlyStopping(monitor="val_auc", mode="max", patience=3, restore_best_weights=True),
            tf.keras.callbacks.ReduceLROnPlateau(monitor="val_loss", factor=0.3, patience=1, min_lr=1e-7),
        ],
    )

    # ---- "after" numbers ----
    best = tf.keras.models.load_model(str(OUT_MODEL), compile=False)
    v_paths, v_labels, new_val = score_split(tf, best, "val")
    choice = choose_threshold(v_labels, new_val)
    threshold = choice["threshold"]
    print(f"\nThreshold chosen on validation (false positives <= {MAX_FALSE_POSITIVE_RATE:.0%}): {threshold}")
    report("AFTER (v2 model) - validation (optimistic: threshold picked here)", v_paths, v_labels, new_val, threshold)
    t_paths, t_labels, new_test = score_split(tf, best, "test")
    report("AFTER (v2 model) - TEST, never seen (use this one)", t_paths, t_labels, new_test, threshold)

    OUT_THRESHOLD.write_text(json.dumps({
        "fake_threshold": threshold,
        "model_file": OUT_MODEL.name,
        "label_order": CLASS_NAMES,
        "max_false_positive_rate": MAX_FALSE_POSITIVE_RATE,
        "validation": choice,
    }, indent=2), encoding="utf-8")

    DRIVE_OUT.mkdir(parents=True, exist_ok=True)
    shutil.copy(OUT_MODEL, DRIVE_OUT / OUT_MODEL.name)
    shutil.copy(OUT_THRESHOLD, DRIVE_OUT / OUT_THRESHOLD.name)
    print("\nSaved to Drive folder:", DRIVE_OUT)
    print(" ", OUT_MODEL.name)
    print(" ", OUT_THRESHOLD.name)


if __name__ == "__main__":
    main()
```

```python
import json, zipfile

src = "/content/work/detect_now_efficientnetb0_faceswap_crops.keras"
dst = "/content/work/detect_now_efficientnetb0_faceswap_crops_patched.keras"

def strip(obj):
    removed = 0
    if isinstance(obj, dict):
        for key in ("input_axes", "output_axes"):
            if key in obj and obj[key] is None:
                del obj[key]; removed += 1
        for v in obj.values(): removed += strip(v)
    elif isinstance(obj, list):
        for v in obj: removed += strip(v)
    return removed

with zipfile.ZipFile(src) as zin, zipfile.ZipFile(dst, "w", zipfile.ZIP_DEFLATED) as zout:
    for item in zin.infolist():
        data = zin.read(item.filename)
        if item.filename == "config.json":
            cfg = json.loads(data)
            print("removed settings:", strip(cfg))
            data = json.dumps(cfg).encode("utf-8")
        zout.writestr(item, data)
print("saved", dst)

!sed -i 's/faceswap_crops\.keras"/faceswap_crops_patched.keras"/' /content/work/train.py
!grep -n "BASE_MODEL =" /content/work/train.py
%cd /content/work
!python train.py
```

```python
import json, os, re, subprocess, sys, zipfile

SRC = "/content/work/detect_now_efficientnetb0_faceswap_crops.keras"
DST = "/content/work/detect_now_efficientnetb0_faceswap_crops_patched.keras"
remove = {"input_axes", "output_axes", "renorm", "renorm_clipping", "renorm_momentum"}

def strip(obj, keys):
    n = 0
    if isinstance(obj, dict):
        for k in [k for k in obj if k in keys]:
            del obj[k]; n += 1
        for v in obj.values(): n += strip(v, keys)
    elif isinstance(obj, list):
        for v in obj: n += strip(v, keys)
    return n

def write_patched():
    with zipfile.ZipFile(SRC) as zin, zipfile.ZipFile(DST, "w", zipfile.ZIP_DEFLATED) as zout:
        for item in zin.infolist():
            data = zin.read(item.filename)
            if item.filename == "config.json":
                cfg = json.loads(data); n = strip(cfg, remove)
                data = json.dumps(cfg).encode("utf-8")
                print("  removed", n, "unknown settings")
            zout.writestr(item, data)

CHECK = ("import os; os.environ['TF_CPP_MIN_LOG_LEVEL']='3'; import tensorflow as tf; "
         f"tf.keras.models.load_model('{DST}', compile=False); print('LOADS_OK')")

loaded = False
for attempt in range(1, 9):
    print(f"attempt {attempt}: removing {sorted(remove)}")
    write_patched()
    r = subprocess.run([sys.executable, "-c", CHECK], capture_output=True, text=True,
                       env={**os.environ, "CUDA_VISIBLE_DEVICES": ""})
    text = r.stdout + r.stderr
    if "LOADS_OK" in text:
        print("Model loads correctly.")
        loaded = True
        break
    new = set(re.findall(r"unexpected keyword argument '([^']+)'", text))
    for chunk in re.findall(r"Unrecognized keyword arguments passed to \w+: \{([^}]*)\}", text):
        new |= set(re.findall(r"'([^']+)'\s*:", chunk))
    new -= remove
    if not new:
        print("Could not fix automatically. Last lines of the error:")
        print("\n".join(text.strip().splitlines()[-12:]))
        break
    print("  also removing:", sorted(new))
    remove |= new

if loaded:
    get_ipython().system("sed -i 's/faceswap_crops\\.keras\"/faceswap_crops_patched.keras\"/' /content/work/train.py")
    get_ipython().system("cd /content/work && python train.py")
```

```python
import os, sys, numpy as np, tensorflow as tf
sys.path.insert(0, "/content/work")
import train

OLD = "/content/work/detect_now_efficientnetb0_faceswap_crops_patched.keras"
NEW = "/content/work/detect_now_efficientnetb0_faceswap_crops_v2.keras"
assert os.path.exists(OLD) and os.path.exists(NEW), "Files missing - the Colab session was reset"

old = tf.keras.models.load_model(OLD, compile=False)
new = tf.keras.models.load_model(NEW, compile=False)

combiners = {
    "old": lambda a, b: a,
    "v2": lambda a, b: b,
    "average": lambda a, b: (a + b) / 2,
    "max": lambda a, b: np.maximum(a, b),
}

for split in ("val", "test"):
    paths, labels, p_old = train.score_split(tf, old, split)
    _, _, p_new = train.score_split(tf, new, split)
    kaggle = np.array([p.split("/")[-1].startswith("kg_") for p in paths])
    for name, mask in (("Kaggle", kaggle), ("inswapper", ~kaggle)):
        if not mask.sum():
            continue
        print(f"\n{name} {split} (n={int(mask.sum())})")
        for label, fn in combiners.items():
            s = fn(p_old, p_new)
            line = f"  {label:8s} AUC {train.subset_auc(labels, s, mask):.3f}"
            if label in ("average", "max"):
                for t in (0.3, 0.4, 0.5):
                    m = train.metrics_at(labels[mask], s[mask], t)
                    line += f" | @{t}: real {m['real_correct']:.0%}, fakes {m['fakes_caught']:.0%}"
            print(line)
```
